```python
## Creating dictionaries

empty_disc = {}
print(type(empty_disc))

empty_dict = dict()
print(type(empty_dict))
```

```python
student = {"name": "Paul", "Age": 23, "Grade": 73}

print(student)
```

```python
## Accessing Dictionaries Elements

student={"name": "Posh", "Age": 25, "Grade": "A"}

print(student["name"])
print(student["Grade"])
print(student.get("Age", "Not Avaliable"))
print(student.get("last_name", "Not Avaliable"))
```

```python
## Dictionary are mutable, meaning you can add, update and delete elements

student["Age"] = 22
print(student)
student["Address"] = "Nigeria"
print(student)
```

```python
del student["Address"]
print(student)
```

```python
keys = student.keys()
print(keys)
values = student.values()
print(values)
items = student.items()
print(items)
```

```python
## Shallow copy
student_copy =student
print(student)
print(student_copy)
```

```python
for keys in student.keys():
    print(keys)
```

```python
for values in student.values():
    print(values)
```

```python
for key,value in student.items():
    print(f"{key}:{value}")
```

```python
students = {
    "student1": {"name": "Kris", "age": 32},
    "student2": {"name": "Chris", "age": 22}
}

print(students)
```

```python
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
for student_id, student_info in students.items():
    print(f"{student_id}{student_info}")
    for key, value in student_info.items():
      print(f"{key}:{value}")
```

```python
## Dictionary Comprehension

squares ={x:x**2 for x in range(10) if x % 2 == 0}
print(squares)
```

```python
## Count the frequency of numbers in an element

numbers = [1,2,2,3,3,3,4,4,5,5,5,5]
frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```
