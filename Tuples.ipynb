```python
tpl = tuple(range(1, 11))
print(f"First Number: {tpl[0]}")
print(f"Middle Number: {tpl[len(tpl) // 2]}")
print(f"Last number: {tpl[-1]}")

print(f"Element from Index 2 to 5: {tpl[2:5]}")
```

```python
numbers=tuple([1,2,3,4,5,6])
print(f"First 3 numbers: {numbers[0:3]}")
print(f"Last 3 element: {numbers[3:6]}")
```
