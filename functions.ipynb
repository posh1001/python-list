```python
def even_or_odd(num):
    if num % 2 == 0:
        print(f"{num} is an even number")
    else:
        print(f"{num} is an odd number")
```

```python
even_or_odd(3)
```

```python
def add(a, b):
    return a + b

result = add(4,5)
print(result)
```

```python
def greet(name="Guest"):
    print(f"Hello {name} Welcome to November")

greet("Kenny")
```

```python
def print_number(*krish):
    for number in krish:
        print(number)

print_number(1,2,3,4,5,6,7,8,9, "posh")
```

```python
## Positional arguments

def print_numbers(*args):
    for number in args:
        print(number)

print_numbers(2,4,6,8,10)
```

```python
## Keywords arguments
## In Keyword argument, all the parameter will be inform of key value pairs

def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_details(name="posh", location="Nigeria", city="Lagos")
```

```python
def print_details(*args, **kwargs):
    for val in args:
        print(f"Positional argument :{val}")

    for key,value in kwargs.items():
        print(f"{key}: {value}")

print_details(1,2,3,4,5,"posh", name="John", age=23, location="lagos")
```
