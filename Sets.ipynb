```python
set1 = {1,2,3,4,5}
set2 = {2,3,1,6,7,8,9,10}

union_set = set1.union(set2)
print(union_set)

intersection_set = set1.intersection(set2)
print(intersection_set)
```

```python
set1 = {1,2,3,4,5,6,7}
set2 = {4,5,6,7,8,9,10}

differnece_set = set1.difference(set2)
update_set = set1.update(set2)
print(differnece_set)
print(update_set)
```

```python
set1.symmetric_difference(set2)
```

```python
## Sets method subset

set1 = {1,2,3,4,5}
set2 = {3,4,5}

print(set1.issubset(set2))
print(set1.issuperset(set2))
```

```python
## Counting unique words

text = "This is a very informative tutorial"
words = text.split()

unique_words = set(words)
print(unique_words)
print(len(unique_words))
```
